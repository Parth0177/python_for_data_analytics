#### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
## Series
## A Pandas series is a 1-d array like object which can store any datatype.
### It is similar to a column in table
import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series\n",series)
print(type(series))
```

```python
### Create a series from dictionary 
data={'a':1,'b':2,'c':3,'d':4}
series_dict= pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
data=['Parth','Ansh','Sargam','Akshita']
roll_No=[11,2,55,13]
pd.Series(data,index=roll_No)
```

```python
### Dataframe using dictionary
data={
  'Name':['Parth','Akshita','SAARGAM'],
  'Age':[22,23,24],
  'city':["Lucknow","Maharajganj","Gonda"]
}

df11=pd.DataFrame(data)
print(df11)
print(type(df11))
```

```python
### Create a data Frame from a list of dictionaries
data=[
  {'Name':'Parth','Age':22,'City':'Lucknow'},
  {'Name':'Sargam','Age':23,'City':'Gonda'},
  {'Name':'Akshita','Age':21,'City':'Gorakhpur'},
]

df=pd.DataFrame(data)
print(df)
```
