### Data Visualization with Matplotlib
Matplotlib is a powerful plottting library for python that enables creation of static, animated and interactive visualiztions. It is widely used for data science and analytics

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

# create a line plot
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Line Plot')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

## cUSTOMISED LINE PLOT
plt.plot(x,y, color='red', marker='o', linestyle='--')
plt.grid(True)
```

```python
## Multiple Plots
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,8,27,64,125]
plt.plot(x,y1, label='y=x^2')
plt.plot(x,y2, label='y=x^3')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Multiple Plots')
plt.legend()
plt.show()
```

```python
### Bar Plot
categories=['A','B','C','D','E']
values=[10,15,7,12,20]
plt.bar(categories,values,color='yellow')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()
```

### Histograms
Histogram is a graphical representation of the distribution of a dataset. It is created by dividing the data into bins and counting the number of data points that fall into each bin. The height of each bar in the histogram represents the frequency of data points in that bin.
