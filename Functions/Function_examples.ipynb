```python
### Temperature Conversion
def convert_temperature(temp,unit):
  """This function converts a temperature from Celsius to Fahrenheit or vice versa."""
  if unit == "C":
    return (temp* 9/5)+32
  elif unit =="F":
    return (temp-32)*5/9
  else:
    return "Invalid unit. Please use 'C' for Celsius or 'F' for Fahrenheit."
```

```python
convert_temperature(32,"C")
```

```python
convert_temperature(89.6,"F")
```

```python
### Password Strength Checker

def check_password(password):
  """this function check whether the password is strong or weak"""
  if(len(password)<8):
    return "Weak Password, Password should be at least 8 characters long"
  elif(not any(char.isdigit() for char in password)):
    return "Weak Password, Password should contain at least one digit"
  elif(not any(char.isupper() for char in password)):
    return "Weak Password, Password should contain at least one uppercase letter"
  elif(not any (char.islower() for char in password )):
    return "Weak Password, Password should contain at least one lowercase letter"
  else:
    return "Strong Password"
```

```python
check_password("Parth._017")
```

```python
check_password("Password it is")
```

```python
check_password("Parth1")
```

```python
### Calculate the total cose of items in a shopping cart
def calculate_cart(cart):
  """This function calculates the total items in a cart"""
  total=0
  for item in cart:
    total=total+item["price"]*item["quantity"]
  return total
  
```

```python
calculate_cart([{"name":"apple","price":10,"quantity":4}
              ,{"name":"banana","price":5,"quantity":6}])
```
