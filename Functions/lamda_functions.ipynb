### Lambda Functions
Lambda functions are small anomynous functions defined using lamda keyword. they can have any number of arguments but only one expression

```python
#syntax
#lambda arguments:expression
```

```python
def add(x,y):
  return x+y
```

```python
add(2,4)
```

```python
addition= lambda x,y:x+y
type(addition)
print(addition(2,4))
```

```python
even= lambda x:x%2==0
print(even(2))
```

```python
odd = lambda x:x%2!=0
print(odd(5))
```

```python
multi= lambda x,y,z:x*y*z
print(multi(2,3,4))
```

```python
power = lambda x,y:x**y
print(power(2,4))
```
