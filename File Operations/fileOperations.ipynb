### File Operations
File handling is a very crucial part of any programming language.Python provides built-in functions and methods to read from and write to files both text and binary

```python
## Read a whole file
with open('file.txt','r')as file:
  content= file.read()
print(content)
```

```python
## Read a file line by line
with open('file.txt','r')as file:
  for line in file:
    print(line)
```

```python
##.strip() method to remove extra new line characters
with open('file.txt','r')as file:
  for line in file:
    print(line.strip())
```

```python
## write to a file(Overwriting)
with open('file.txt','w') as file:
  file.write('This is a new Line \n')
  file.write('Jai Mahakal \n')
```

```python
## Write to a file(Appending)without overwriting
with open('file.txt','a') as file:
  file.write('This is an appended line \n')
  file.write('Hare Krishna \n')
```

```python
### Wrtiing a list of lines to a file
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']
with open('file.txt','a') as file:
  file.writelines(lines)
```

```python
### Binary Files
# Writing to a binary file
data= b'\x00\x01\x02\x03\x04'
with open('binaryfile.bin','wb') as file:
  file.write(data)
```

```python
# Reading to a binary file
with open('binaryfile.bin','rb') as file:
  content= file.read()
print(content)
```

```python
### Read the content from a source text file and write it to a new text file
with open('file.txt','r') as source_file:
  content= source_file.read()
  
with open('newfile.txt','w') as dest_file:
  dest_file.write(content)
```

```python
### Read a text file and count the number of lines, words, and characters
with open('file.txt','r') as file:
  content=file.read()
  lines=content.splitlines()
  words=content.split()
  num_lines=len(lines)
  num_words=len(words)
  num_characters=len(content)
print(f'Number of lines: {num_lines}')
print(f'Number of words: {num_words}')
print(f'Number of characters: {num_characters}')
```
